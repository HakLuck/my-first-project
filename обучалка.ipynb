```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
data={
    'Тема': ['Python','SQL','Pandas',"Визуализация","Мат. статистика"],
    "Прогресс(%)": [90, 75, 40, 15, 5]
}
df = pd.DataFrame(data)
plt.figure(figsize=(8, 4))
sns.barplot(x='Прогресс(%)',y="Тема", data=df, palette="viridis")
plt.xlim(0, 100)
plt.grid(axis="x", linestyle="--",alpha=0.7)
plt.show()
```
